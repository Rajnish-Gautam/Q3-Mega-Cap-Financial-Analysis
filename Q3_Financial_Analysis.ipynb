```python
def get_company_facts(cik):
    url = f"https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json"
    
    response = requests.get(url, headers=headers)
    response.raise_for_status()
    
    return response.json()
```

```python
def get_q3_financials(company_name, cik):
    
    # Get SEC Company Facts
    facts = get_company_facts(cik)
    us_gaap = facts['facts']['us-gaap']
    
    # -----------------------------
    # 1. Find Revenue
    # -----------------------------
    
    revenue_tags = [
        'RevenueFromContractWithCustomerExcludingAssessedTax',
        'Revenues',
        'SalesRevenueNet',
        'RevenueFromContractWithCustomerIncludingAssessedTax'
    ]
    
    revenue_df = None
    revenue_tag_used = None
    
    for tag in revenue_tags:
        
        if tag not in us_gaap:
            continue
        
        try:
            units = us_gaap[tag]['units']
            
            if 'USD' not in units:
                continue
            
            temp = pd.DataFrame(units['USD'])
            
            # Look specifically for standalone Q3 2025
            q3 = temp[
                (temp['start'] == '2025-07-01') &
                (temp['end'] == '2025-09-30') &
                (temp['fp'] == 'Q3') &
                (temp['form'] == '10-Q')
            ]
            
            if len(q3) > 0:
                revenue_df = q3.copy()
                revenue_tag_used = tag
                break
                
        except Exception:
            continue
    
    
    # -----------------------------
    # 2. Find Net Income
    # -----------------------------
    
    if 'NetIncomeLoss' not in us_gaap:
        raise ValueError(f"{company_name}: NetIncomeLoss tag not found")
    
    net_income_units = us_gaap['NetIncomeLoss']['units']
    
    if 'USD' not in net_income_units:
        raise ValueError(f"{company_name}: NetIncomeLoss does not have USD units")
    
    net_income_df = pd.DataFrame(net_income_units['USD'])
    
    net_income_q3 = net_income_df[
        (net_income_df['start'] == '2025-07-01') &
        (net_income_df['end'] == '2025-09-30') &
        (net_income_df['fp'] == 'Q3') &
        (net_income_df['form'] == '10-Q')
    ].copy()
    
    
    # -----------------------------
    # 3. Check that data was found
    # -----------------------------
    
    if revenue_df is None or len(revenue_df) == 0:
        raise ValueError(f"{company_name}: Q3 2025 revenue not found")
    
    if len(net_income_q3) == 0:
        raise ValueError(f"{company_name}: Q3 2025 net income not found")
    
    
    # -----------------------------
    # 4. Take the latest filing
    # -----------------------------
    
    revenue_df = revenue_df.sort_values('filed').tail(1)
    net_income_q3 = net_income_q3.sort_values('filed').tail(1)
    
    
    # -----------------------------
    # 5. Extract values
    # -----------------------------
    
    revenue = revenue_df.iloc[0]['val']
    net_income = net_income_q3.iloc[0]['val']
    
    net_profit_margin = (net_income / revenue) * 100
    
    
    # -----------------------------
    # 6. Return result
    # -----------------------------
    
    return {
        'name': company_name,
        'Revenue': revenue,
        'Net_Income': net_income,
        'Net_Profit_Margin': net_profit_margin,
        'Revenue_Tag': revenue_tag_used
    }
```

```python
def get_q3_financials(company_name, cik):
    
    # Get SEC Company Facts
    facts = get_company_facts(cik)
    us_gaap = facts['facts']['us-gaap']
    
    # -----------------------------
    # 1. Find Revenue
    # -----------------------------
    
    revenue_tags = [
        'RevenueFromContractWithCustomerExcludingAssessedTax',
        'Revenues',
        'SalesRevenueNet',
        'RevenueFromContractWithCustomerIncludingAssessedTax'
    ]
    
    revenue_df = None
    revenue_tag_used = None
    
    for tag in revenue_tags:
        
        if tag not in us_gaap:
            continue
        
        if 'USD' not in us_gaap[tag]['units']:
            continue
        
        temp = pd.DataFrame(us_gaap[tag]['units']['USD'])
        
        q3 = temp[
            (temp['fy'] == 2025) &
            (temp['fp'] == 'Q3') &
            (temp['form'] == '10-Q')
        ].copy()
        
        if len(q3) > 0:
            q3['start'] = pd.to_datetime(q3['start'])
            q3['end'] = pd.to_datetime(q3['end'])
            
            # Calculate reporting-period length
            q3['days'] = (q3['end'] - q3['start']).dt.days
            
            # Shortest period = standalone quarter
            revenue_df = q3.sort_values(
                ['days', 'filed']
            ).tail(1)
            
            # Correct the selection: shortest period, latest filing
            min_days = q3['days'].min()
            revenue_df = q3[q3['days'] == min_days].sort_values('filed').tail(1)
            
            revenue_tag_used = tag
            break
    
    
    # -----------------------------
    # 2. Find Net Income
    # -----------------------------
    
    if 'NetIncomeLoss' not in us_gaap:
        raise ValueError(f"{company_name}: NetIncomeLoss tag not found")
    
    if 'USD' not in us_gaap['NetIncomeLoss']['units']:
        raise ValueError(f"{company_name}: NetIncomeLoss does not have USD units")
    
    net_income_df = pd.DataFrame(
        us_gaap['NetIncomeLoss']['units']['USD']
    )
    
    net_income_q3 = net_income_df[
        (net_income_df['fy'] == 2025) &
        (net_income_df['fp'] == 'Q3') &
        (net_income_df['form'] == '10-Q')
    ].copy()
    
    if len(net_income_q3) == 0:
        raise ValueError(f"{company_name}: Q3 FY2025 net income not found")
    
    net_income_q3['start'] = pd.to_datetime(net_income_q3['start'])
    net_income_q3['end'] = pd.to_datetime(net_income_q3['end'])
    net_income_q3['days'] = (
        net_income_q3['end'] - net_income_q3['start']
    ).dt.days
    
    # Shortest period = standalone quarter
    min_days = net_income_q3['days'].min()
    
    net_income_q3 = net_income_q3[
        net_income_q3['days'] == min_days
    ].sort_values('filed').tail(1)
    
    
    # -----------------------------
    # 3. Check Revenue
    # -----------------------------
    
    if revenue_df is None or len(revenue_df) == 0:
        raise ValueError(f"{company_name}: Q3 FY2025 revenue not found")
    
    
    # -----------------------------
    # 4. Extract values
    # -----------------------------
    
    revenue = revenue_df.iloc[0]['val']
    net_income = net_income_q3.iloc[0]['val']
    
    net_profit_margin = (net_income / revenue) * 100
    
    
    # -----------------------------
    # 5. Return result
    # -----------------------------
    
    return {
        'name': company_name,
        'Revenue': revenue,
        'Net_Income': net_income,
        'Net_Profit_Margin': net_profit_margin,
        'Revenue_Tag': revenue_tag_used,
        'Period_Start': revenue_df.iloc[0]['start'],
        'Period_End': revenue_df.iloc[0]['end']
    }
```

```python
def analyze_companies(ciks):
    
    results = []
    
    for company, cik in ciks.items():
        
        try:
            # Get company facts from SEC
            facts_url = f"https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json"
            
            response = requests.get(
                facts_url,
                headers=headers
            )
            
            response.raise_for_status()
            
            facts = response.json()
            
            # Find revenue tags
            revenue_tags = [
                'RevenueFromContractWithCustomerExcludingAssessedTax',
                'Revenues'
            ]
            
            revenue_df = None
            revenue_tag_used = None
            
            for tag in revenue_tags:
                try:
                    data = facts['facts']['us-gaap'][tag]['units']['USD']
                    temp = pd.DataFrame(data)
                    
                    if not temp.empty:
                        revenue_df = temp
                        revenue_tag_used = tag
                        break
                        
                except KeyError:
                    continue
            
            if revenue_df is None:
                raise ValueError("Revenue not found")
            
            # Net income
            net_income_data = facts['facts']['us-gaap']['NetIncomeLoss']['units']['USD']
            net_income_df = pd.DataFrame(net_income_data)
            
            # Convert dates
            revenue_df['end'] = pd.to_datetime(revenue_df['end'])
            net_income_df['end'] = pd.to_datetime(net_income_df['end'])
            
            if 'start' in revenue_df.columns:
                revenue_df['start'] = pd.to_datetime(revenue_df['start'])
            
            if 'start' in net_income_df.columns:
                net_income_df['start'] = pd.to_datetime(net_income_df['start'])
            
            # Find Q3 records
            revenue_q3 = revenue_df[
                (revenue_df['fp'] == 'Q3') &
                (revenue_df['form'] == '10-Q')
            ].copy()
            
            net_income_q3 = net_income_df[
                (net_income_df['fp'] == 'Q3') &
                (net_income_df['form'] == '10-Q')
            ].copy()
            
            # Keep recent records
            revenue_q3 = revenue_q3.sort_values('filed')
            net_income_q3 = net_income_q3.sort_values('filed')
            
            # Select latest Q3 record
            revenue_latest = revenue_q3.iloc[-1]
            net_income_latest = net_income_q3.iloc[-1]
            
            # Make sure revenue and income belong to same quarter
            revenue_start = revenue_latest['start']
            revenue_end = revenue_latest['end']
            
            matching_income = net_income_q3[
                (net_income_q3['start'] == revenue_start) &
                (net_income_q3['end'] == revenue_end)
            ]
            
            if matching_income.empty:
                raise ValueError("Matching Q3 net income not found")
            
            net_income_latest = matching_income.iloc[-1]
            
            revenue = revenue_latest['val']
            net_income = net_income_latest['val']
            
            margin = (net_income / revenue) * 100
            
            results.append({
                'name': company,
                'Revenue': revenue,
                'Net_Income': net_income,
                'Net_Profit_Margin': margin,
                'Revenue_Tag': revenue_tag_used,
                'Period_Start': revenue_start,
                'Period_End': revenue_end
            })
            
            print(f"✓ {company}")
            
        except Exception as e:
            print(f"✗ {company}: {e}")
    
    return pd.DataFrame(results)
```

```python
def analyze_companies(ciks, target_fy=2025):

    results = []

    for company, cik in ciks.items():

        try:
            # -----------------------------------
            # 1. Get SEC Company Facts
            # -----------------------------------
            facts_url = f"https://data.sec.gov/api/xbrl/companyfacts/CIK{cik}.json"

            response = requests.get(
                facts_url,
                headers=headers
            )

            response.raise_for_status()

            facts = response.json()

            us_gaap = facts['facts']['us-gaap']

            # -----------------------------------
            # 2. Find the correct Revenue tag
            # -----------------------------------
            revenue_tags = [
                'RevenueFromContractWithCustomerExcludingAssessedTax',
                'Revenues'
            ]

            usable_revenue_df = None
            revenue_tag_used = None

            for tag in revenue_tags:

                if tag not in us_gaap:
                    continue

                try:
                    revenue_data = us_gaap[tag]['units']['USD']

                    temp_revenue = pd.DataFrame(
                        revenue_data
                    )

                    if temp_revenue.empty:
                        continue

                    # Convert dates
                    temp_revenue['start'] = pd.to_datetime(
                        temp_revenue['start'],
                        errors='coerce'
                    )

                    temp_revenue['end'] = pd.to_datetime(
                        temp_revenue['end'],
                        errors='coerce'
                    )

                    # -----------------------------------
                    # Keep target FY + Q3 + 10-Q
                    # -----------------------------------
                    temp_q3 = temp_revenue[
                        (temp_revenue['fy'] == target_fy) &
                        (temp_revenue['fp'] == 'Q3') &
                        (temp_revenue['form'] == '10-Q')
                    ].copy()

                    if temp_q3.empty:
                        continue

                    # -----------------------------------
                    # Calculate duration
                    # -----------------------------------
                    temp_q3['duration_days'] = (
                        temp_q3['end'] -
                        temp_q3['start']
                    ).dt.days

                    # -----------------------------------
                    # Keep standalone quarterly values
                    # -----------------------------------
                    temp_actual = temp_q3[
                        temp_q3['duration_days'].between(
                            80, 100
                        )
                    ].copy()

                    if temp_actual.empty:
                        continue

                    usable_revenue_df = temp_actual
                    revenue_tag_used = tag

                    break

                except (KeyError, TypeError):
                    continue

            # -----------------------------------
            # 3. Check Revenue
            # -----------------------------------
            if usable_revenue_df is None:
                raise ValueError(
                    f"Q3 FY{target_fy} revenue not found"
                )

            # -----------------------------------
            # 4. Select Q3 Revenue
            # -----------------------------------
            usable_revenue_df = usable_revenue_df.sort_values(
                ['end', 'filed']
            )

            revenue_latest = usable_revenue_df.iloc[-1]

            revenue_start = revenue_latest['start']
            revenue_end = revenue_latest['end']

            # -----------------------------------
            # 5. Get Net Income
            # -----------------------------------
            if 'NetIncomeLoss' not in us_gaap:
                raise ValueError(
                    "Net Income data not found"
                )

            net_income_data = (
                us_gaap['NetIncomeLoss']
                ['units']['USD']
            )

            net_income_df = pd.DataFrame(
                net_income_data
            )

            # Convert dates
            net_income_df['start'] = pd.to_datetime(
                net_income_df['start'],
                errors='coerce'
            )

            net_income_df['end'] = pd.to_datetime(
                net_income_df['end'],
                errors='coerce'
            )

            # -----------------------------------
            # 6. Keep target FY + Q3 + 10-Q
            # -----------------------------------
            net_income_q3 = net_income_df[
                (net_income_df['fy'] == target_fy) &
                (net_income_df['fp'] == 'Q3') &
                (net_income_df['form'] == '10-Q')
            ].copy()

            if net_income_q3.empty:
                raise ValueError(
                    f"Q3 FY{target_fy} net income not found"
                )

            # -----------------------------------
            # 7. Calculate duration
            # -----------------------------------
            net_income_q3['duration_days'] = (
                net_income_q3['end'] -
                net_income_q3['start']
            ).dt.days

            net_income_actual = net_income_q3[
                net_income_q3['duration_days'].between(
                    80, 100
                )
            ].copy()

            if net_income_actual.empty:
                raise ValueError(
                    "Standalone quarterly net income not found"
                )

            # -----------------------------------
            # 8. Match Net Income to Revenue
            # -----------------------------------
            matching_income = net_income_actual[
                (net_income_actual['start'] == revenue_start) &
                (net_income_actual['end'] == revenue_end)
            ].copy()

            if matching_income.empty:
                raise ValueError(
                    "Matching quarterly net income not found"
                )

            matching_income = matching_income.sort_values(
                'filed'
            )

            net_income_latest = matching_income.iloc[-1]

            # -----------------------------------
            # 9. Extract values
            # -----------------------------------
            revenue = revenue_latest['val']
            net_income = net_income_latest['val']

            # -----------------------------------
            # 10. Calculate Net Profit Margin
            # -----------------------------------
            margin = (
                net_income / revenue
            ) * 100

            # -----------------------------------
            # 11. Store result
            # -----------------------------------
            results.append({
                'name': company,
                'Revenue': revenue,
                'Net_Income': net_income,
                'Net_Profit_Margin': margin,
                'Revenue_Tag': revenue_tag_used,
                'Period_Start': revenue_start,
                'Period_End': revenue_end
            })

            print(f"✓ {company}")

        except Exception as e:

            print(f"✗ {company}: {e}")

    return pd.DataFrame(results)
```

```python
# -----------------------------------------
# WEIGHTS
# -----------------------------------------

revenue_weight = 0.30
net_income_weight = 0.30
margin_weight = 0.40


# -----------------------------------------
# CONVERT RANKS INTO SCORES
# Rank 1 = highest score
# -----------------------------------------

number_of_companies = len(automated_data)

automated_data['Revenue_Score'] = (
    (number_of_companies - automated_data['Revenue_Rank'] + 1)
    / number_of_companies
)

automated_data['Net_Income_Score'] = (
    (number_of_companies - automated_data['Net_Income_Rank'] + 1)
    / number_of_companies
)

automated_data['Margin_Score'] = (
    (number_of_companies - automated_data['Margin_Rank'] + 1)
    / number_of_companies
)


# -----------------------------------------
# WEIGHTED SCORE
# -----------------------------------------

automated_data['Weighted_Score'] = (
    automated_data['Revenue_Score'] * revenue_weight
    + automated_data['Net_Income_Score'] * net_income_weight
    + automated_data['Margin_Score'] * margin_weight
)


# -----------------------------------------
# WEIGHTED RANK
# -----------------------------------------

automated_data['Weighted_Rank'] = (
    automated_data['Weighted_Score']
    .rank(
        ascending=False,
        method='min'
    )
    .astype(int)
)


# -----------------------------------------
# DISPLAY RESULTS
# -----------------------------------------

weighted_results = automated_data.sort_values(
    'Weighted_Rank'
).reset_index(drop=True)

weighted_results[
    [
        'name',
        'Revenue_Bn',
        'Net_Income_Bn',
        'Net_Profit_Margin',
        'Revenue_Score',
        'Net_Income_Score',
        'Margin_Score',
        'Weighted_Score',
        'Weighted_Rank'
    ]
]
```

```python
import pandas as pd
import requests

# ============================================================
# SEC HEADERS
# ============================================================

headers = {
    "User-Agent": "Your Full Name your_email@example.com"
}


# ============================================================
# GET SEC COMPANY FACTS
# ============================================================

def get_company_facts(cik):

    cik = str(cik).zfill(10)

    url = (
        f"https://data.sec.gov/api/xbrl/companyfacts/"
        f"CIK{cik}.json"
    )

    response = requests.get(
        url,
        headers=headers
    )

    response.raise_for_status()

    return response.json()


# ============================================================
# GET XBRL TAG
# ============================================================

def get_xbrl_tag(facts, tag):

    try:

        data = facts["facts"]["us-gaap"][tag]["units"]

        if "USD" not in data:
            return None

        df = pd.DataFrame(data["USD"])

        if df.empty:
            return None

        for col in ["start", "end", "filed"]:

            if col in df.columns:

                df[col] = pd.to_datetime(
                    df[col],
                    errors="coerce"
                )

        df["val"] = pd.to_numeric(
            df["val"],
            errors="coerce"
        )

        df["duration_days"] = (
            df["end"] -
            df["start"]
        ).dt.days

        return df

    except Exception:

        return None


# ============================================================
# REVENUE
# ============================================================

def get_revenue(facts):

    tags = [
        "RevenueFromContractWithCustomerExcludingAssessedTax",
        "Revenues"
    ]

    for tag in tags:

        df = get_xbrl_tag(
            facts,
            tag
        )

        if df is not None and not df.empty:

            df["Revenue_Tag"] = tag

            return df

    return None


# ============================================================
# NET INCOME
# ============================================================

def get_net_income(facts):

    tags = [
        "ProfitLoss",
        "NetIncomeLoss"
    ]

    for tag in tags:

        df = get_xbrl_tag(
            facts,
            tag
        )

        if df is not None and not df.empty:

            df["Net_Income_Tag"] = tag

            return df

    return None


# ============================================================
# CLEAN QUARTERLY DATA
# ============================================================

def clean_quarters(df):

    if df is None or df.empty:

        return pd.DataFrame()

    df = df.copy()

    # 10-Q only
    df = df[
        df["form"]
        .astype(str)
        .str.upper()
        .eq("10-Q")
    ]

    # Valid dates
    df = df[
        df["start"].notna() &
        df["end"].notna()
    ]

    # Quarterly duration
    df = df[
        df["duration_days"].between(
            80,
            100
        )
    ]

    # Remove exact duplicates
    df = df.drop_duplicates(
        subset=[
            "start",
            "end",
            "val"
        ]
    )

    return df.reset_index(drop=True)


# ============================================================
# FIND PRIOR YEAR — DATE METHOD
# ============================================================

def find_prior_year_by_date(
    df,
    current_start,
    current_end
):

    df = clean_quarters(df)

    if df.empty:

        return None

    current_start = pd.Timestamp(
        current_start
    )

    current_end = pd.Timestamp(
        current_end
    )

    target_start = (
        current_start -
        pd.DateOffset(years=1)
    )

    target_end = (
        current_end -
        pd.DateOffset(years=1)
    )

    df["start_diff"] = (
        (
            df["start"] -
            target_start
        )
        .abs()
        .dt.days
    )

    df["end_diff"] = (
        (
            df["end"] -
            target_end
        )
        .abs()
        .dt.days
    )

    df["total_diff"] = (
        df["start_diff"] +
        df["end_diff"]
    )

    df = df.sort_values(
        [
            "total_diff",
            "filed"
        ]
    )

    best = df.iloc[0]

    if best["total_diff"] <= 45:

        return best

    return None


# ============================================================
# FIND PRIOR YEAR — FISCAL QUARTER METHOD
#
# This is the fallback for companies such as NVIDIA
# whose fiscal calendar does not align cleanly with
# calendar dates.
# ============================================================

def find_prior_year_by_fiscal_quarter(
    df,
    current_end,
    current_fy,
    current_fp
):

    df = clean_quarters(df)

    if df.empty:

        return None

    current_fy = pd.to_numeric(
        current_fy,
        errors="coerce"
    )

    current_fp = str(
        current_fp
    ).upper()

    if pd.isna(current_fy):

        return None

    # Prior fiscal year, same fiscal quarter
    target_fy = current_fy - 1

    candidates = df[
        (pd.to_numeric(
            df["fy"],
            errors="coerce"
        ) == target_fy)
        &
        (
            df["fp"]
            .astype(str)
            .str.upper()
            == current_fp
        )
    ].copy()

    if candidates.empty:

        return None

    # If multiple observations exist,
    # choose latest filing
    candidates = candidates.sort_values(
        "filed"
    )

    return candidates.iloc[-1]


# ============================================================
# MAIN YOY AUTOMATION
# ============================================================

yoy_results = []


for _, row in automated_data.iterrows():

    company = row["name"]

    try:

        print(
            f"\nProcessing {company}..."
        )

        # ----------------------------------------------------
        # CIK
        # ----------------------------------------------------

        cik = ciks[company]

        facts = get_company_facts(
            cik
        )


        # ----------------------------------------------------
        # DATA
        # ----------------------------------------------------

        revenue_df = get_revenue(
            facts
        )

        net_income_df = get_net_income(
            facts
        )


        if revenue_df is None:

            print(
                f"✗ {company}: revenue data unavailable"
            )

            continue


        if net_income_df is None:

            print(
                f"✗ {company}: net income data unavailable"
            )

            continue


        # ----------------------------------------------------
        # CURRENT VALUES
        #
        # These are already validated by automated_data.
        # ----------------------------------------------------

        current_revenue = float(
            row["Revenue"]
        )

        current_net_income = float(
            row["Net_Income"]
        )

        current_start = pd.Timestamp(
            row["Period_Start"]
        )

        current_end = pd.Timestamp(
            row["Period_End"]
        )


        # ----------------------------------------------------
        # TRY TO GET FISCAL YEAR / QUARTER FROM SEC
        # ----------------------------------------------------

        revenue_clean = clean_quarters(
            revenue_df
        )

        current_fiscal_row = None


        if not revenue_clean.empty:

            revenue_clean["end_diff"] = (
                (
                    revenue_clean["end"]
                    - current_end
                )
                .abs()
                .dt.days
            )

            close_rows = revenue_clean[
                revenue_clean["end_diff"] <= 15
            ].sort_values(
                [
                    "end_diff",
                    "filed"
                ]
            )

            if not close_rows.empty:

                current_fiscal_row = (
                    close_rows.iloc[-1]
                )


        # ----------------------------------------------------
        # PRIOR-YEAR REVENUE
        # ----------------------------------------------------

        previous_revenue = (
            find_prior_year_by_date(
                revenue_df,
                current_start,
                current_end
            )
        )


        # ----------------------------------------------------
        # FISCAL FALLBACK
        # ----------------------------------------------------

        if (
            previous_revenue is None
            and
            current_fiscal_row is not None
        ):

            previous_revenue = (
                find_prior_year_by_fiscal_quarter(
                    revenue_df,
                    current_end,
                    current_fiscal_row["fy"],
                    current_fiscal_row["fp"]
                )
            )


        if previous_revenue is None:

            print(
                f"✗ {company}: prior-year revenue not found"
            )

            continue


        # ----------------------------------------------------
        # PRIOR-YEAR NET INCOME
        # ----------------------------------------------------

        previous_net_income = (
            find_prior_year_by_date(
                net_income_df,
                current_start,
                current_end
            )
        )


        # ----------------------------------------------------
        # FISCAL FALLBACK FOR NET INCOME
        # ----------------------------------------------------

        if (
            previous_net_income is None
            and
            current_fiscal_row is not None
        ):

            previous_net_income = (
                find_prior_year_by_fiscal_quarter(
                    net_income_df,
                    current_end,
                    current_fiscal_row["fy"],
                    current_fiscal_row["fp"]
                )
            )


        if previous_net_income is None:

            print(
                f"✗ {company}: prior-year net income not found"
            )

            continue


        # ----------------------------------------------------
        # PRIOR YEAR VALUES
        # ----------------------------------------------------

        previous_revenue_value = float(
            previous_revenue["val"]
        )

        previous_net_income_value = float(
            previous_net_income["val"]
        )


        # ----------------------------------------------------
        # MARGINS
        # ----------------------------------------------------

        current_margin = (
            current_net_income /
            current_revenue
        ) * 100


        previous_margin = (
            previous_net_income_value /
            previous_revenue_value
        ) * 100


        # ----------------------------------------------------
        # YOY
        # ----------------------------------------------------

        revenue_yoy = (
            (
                current_revenue /
                previous_revenue_value
            ) - 1
        ) * 100


        net_income_yoy = (
            (
                current_net_income /
                previous_net_income_value
            ) - 1
        ) * 100


        margin_change = (
            current_margin -
            previous_margin
        )


        # ----------------------------------------------------
        # STORE
        # ----------------------------------------------------

        yoy_results.append({

            "name": company,

            "Current_Revenue_Bn":
                current_revenue / 1e9,

            "Previous_Revenue_Bn":
                previous_revenue_value / 1e9,

            "Revenue_YOY_%":
                revenue_yoy,

            "Current_Net_Income_Bn":
                current_net_income / 1e9,

            "Previous_Net_Income_Bn":
                previous_net_income_value / 1e9,

            "Net_Income_YOY_%":
                net_income_yoy,

            "Current_Margin_%":
                current_margin,

            "Previous_Margin_%":
                previous_margin,

            "Margin_Change_Points":
                margin_change,

            "Current_Period_Start":
                current_start,

            "Current_Period_End":
                current_end,

            "Previous_Period_Start":
                previous_revenue["start"],

            "Previous_Period_End":
                previous_revenue["end"]
        })


        print(
            f"✓ {company}"
        )


    except Exception as e:

        print(
            f"✗ {company}: {e}"
        )


# ============================================================
# CREATE YOY DATAFRAME
# ============================================================

yoy_data = pd.DataFrame(
    yoy_results
)


# ============================================================
# ROUND NUMBERS
# ============================================================

if not yoy_data.empty:

    numeric_columns = [
        "Current_Revenue_Bn",
        "Previous_Revenue_Bn",
        "Revenue_YOY_%",
        "Current_Net_Income_Bn",
        "Previous_Net_Income_Bn",
        "Net_Income_YOY_%",
        "Current_Margin_%",
        "Previous_Margin_%",
        "Margin_Change_Points"
    ]

    yoy_data[numeric_columns] = (
        yoy_data[numeric_columns]
        .round(2)
    )


# ============================================================
# DISPLAY
# ============================================================

yoy_data
```

```python
import pandas as pd
import requests

# ============================================================
# SEC HEADERS
# ============================================================

headers = {
    "User-Agent": "Your Full Name your_email@example.com"
}


# ============================================================
# GET SEC COMPANY FACTS
# ============================================================

def get_company_facts(cik):

    cik = str(cik).zfill(10)

    url = (
        f"https://data.sec.gov/api/xbrl/companyfacts/"
        f"CIK{cik}.json"
    )

    response = requests.get(
        url,
        headers=headers
    )

    response.raise_for_status()

    return response.json()


# ============================================================
# GET XBRL TAG
# ============================================================

def get_xbrl_tag(facts, tag):

    try:

        data = facts["facts"]["us-gaap"][tag]["units"]

        if "USD" not in data:
            return None

        df = pd.DataFrame(
            data["USD"]
        )

        if df.empty:
            return None

        # Dates
        for col in ["start", "end", "filed"]:

            if col in df.columns:

                df[col] = pd.to_datetime(
                    df[col],
                    errors="coerce"
                )

        # Numeric value
        df["val"] = pd.to_numeric(
            df["val"],
            errors="coerce"
        )

        # Duration
        if (
            "start" in df.columns
            and
            "end" in df.columns
        ):

            df["duration_days"] = (
                df["end"] -
                df["start"]
            ).dt.days

        return df

    except Exception:

        return None


# ============================================================
# REVENUE
# ============================================================

def get_revenue(facts):

    revenue_tags = [
        "RevenueFromContractWithCustomerExcludingAssessedTax",
        "Revenues"
    ]

    for tag in revenue_tags:

        df = get_xbrl_tag(
            facts,
            tag
        )

        if (
            df is not None
            and
            not df.empty
        ):

            df["Revenue_Tag"] = tag

            return df

    return None


# ============================================================
# NET INCOME
# ============================================================

def get_net_income(facts):

    net_income_tags = [
        "ProfitLoss",
        "NetIncomeLoss"
    ]

    for tag in net_income_tags:

        df = get_xbrl_tag(
            facts,
            tag
        )

        if (
            df is not None
            and
            not df.empty
        ):

            df["Net_Income_Tag"] = tag

            return df

    return None


# ============================================================
# CLEAN QUARTERLY DATA
# ============================================================

def clean_quarters(df):

    if df is None or df.empty:

        return pd.DataFrame()

    df = df.copy()

    # --------------------------------------------
    # Only 10-Q
    # --------------------------------------------

    if "form" in df.columns:

        df = df[
            df["form"]
            .astype(str)
            .str.upper()
            .eq("10-Q")
        ]

    # --------------------------------------------
    # Valid dates
    # --------------------------------------------

    df = df[
        df["start"].notna()
        &
        df["end"].notna()
    ]

    # --------------------------------------------
    # Quarterly duration
    # --------------------------------------------

    df = df[
        df["duration_days"].between(
            80,
            100
        )
    ]

    # --------------------------------------------
    # Valid values
    # --------------------------------------------

    df = df[
        df["val"].notna()
    ]

    return df.reset_index(
        drop=True
    )


# ============================================================
# FIND PRIOR-YEAR QUARTER
#
# Key improvement:
#
# We compare the ACTUAL PERIOD DATES rather than FY labels.
#
# This handles NVIDIA's comparative XBRL observations.
# ============================================================

def find_prior_year_quarter(
    df,
    current_start,
    current_end
):

    df = clean_quarters(
        df
    )

    if df.empty:

        return None

    current_start = pd.Timestamp(
        current_start
    )

    current_end = pd.Timestamp(
        current_end
    )

    # --------------------------------------------------------
    # Target period = approximately one year earlier
    # --------------------------------------------------------

    target_start = (
        current_start -
        pd.DateOffset(
            years=1
        )
    )

    target_end = (
        current_end -
        pd.DateOffset(
            years=1
        )
    )

    # --------------------------------------------------------
    # Calculate differences
    # --------------------------------------------------------

    df["start_diff"] = (
        (
            df["start"] -
            target_start
        )
        .abs()
        .dt.days
    )

    df["end_diff"] = (
        (
            df["end"] -
            target_end
        )
        .abs()
        .dt.days
    )

    df["total_diff"] = (
        df["start_diff"] +
        df["end_diff"]
    )

    # --------------------------------------------------------
    # Sort by closest actual period
    # --------------------------------------------------------

    df = df.sort_values(
        [
            "total_diff",
            "filed"
        ]
    )

    # --------------------------------------------------------
    # IMPORTANT:
    #
    # Don't require the fiscal-year label to match.
    # NVIDIA's comparative observation demonstrates why.
    # --------------------------------------------------------

    best = df.iloc[0]

    if best["total_diff"] <= 45:

        return best

    return None


# ============================================================
# MAIN YOY AUTOMATION
# ============================================================

yoy_results = []


for _, row in automated_data.iterrows():

    company = row["name"]

    try:

        print(
            f"\nProcessing {company}..."
        )

        # ----------------------------------------------------
        # CIK
        # ----------------------------------------------------

        cik = ciks[
            company
        ]

        facts = get_company_facts(
            cik
        )


        # ----------------------------------------------------
        # GET REVENUE
        # ----------------------------------------------------

        revenue_df = get_revenue(
            facts
        )


        # ----------------------------------------------------
        # GET NET INCOME
        # ----------------------------------------------------

        net_income_df = get_net_income(
            facts
        )


        if revenue_df is None:

            print(
                f"✗ {company}: revenue unavailable"
            )

            continue


        if net_income_df is None:

            print(
                f"✗ {company}: net income unavailable"
            )

            continue


        # ----------------------------------------------------
        # CURRENT QUARTER
        #
        # Use the already validated automated_data.
        # ----------------------------------------------------

        current_start = pd.Timestamp(
            row["Period_Start"]
        )

        current_end = pd.Timestamp(
            row["Period_End"]
        )


        current_revenue = float(
            row["Revenue"]
        )

        current_net_income = float(
            row["Net_Income"]
        )


        # ----------------------------------------------------
        # PRIOR-YEAR REVENUE
        # ----------------------------------------------------

        previous_revenue = (
            find_prior_year_quarter(
                revenue_df,
                current_start,
                current_end
            )
        )


        if previous_revenue is None:

            print(
                f"✗ {company}: "
                f"prior-year revenue not found"
            )

            continue


        # ----------------------------------------------------
        # PRIOR-YEAR NET INCOME
        # ----------------------------------------------------

        previous_net_income = (
            find_prior_year_quarter(
                net_income_df,
                current_start,
                current_end
            )
        )


        if previous_net_income is None:

            print(
                f"✗ {company}: "
                f"prior-year net income not found"
            )

            continue


        # ----------------------------------------------------
        # PRIOR-YEAR VALUES
        # ----------------------------------------------------

        previous_revenue_value = float(
            previous_revenue["val"]
        )

        previous_net_income_value = float(
            previous_net_income["val"]
        )


        # ----------------------------------------------------
        # CURRENT MARGIN
        # ----------------------------------------------------

        current_margin = (
            current_net_income /
            current_revenue
        ) * 100


        # ----------------------------------------------------
        # PRIOR-YEAR MARGIN
        # ----------------------------------------------------

        previous_margin = (
            previous_net_income_value /
            previous_revenue_value
        ) * 100


        # ----------------------------------------------------
        # REVENUE YOY
        # ----------------------------------------------------

        revenue_yoy = (
            (
                current_revenue /
                previous_revenue_value
            )
            - 1
        ) * 100


        # ----------------------------------------------------
        # NET INCOME YOY
        # ----------------------------------------------------

        net_income_yoy = (
            (
                current_net_income /
                previous_net_income_value
            )
            - 1
        ) * 100


        # ----------------------------------------------------
        # MARGIN CHANGE
        # ----------------------------------------------------

        margin_change = (
            current_margin -
            previous_margin
        )


        # ----------------------------------------------------
        # SAVE RESULT
        # ----------------------------------------------------

        yoy_results.append({

            "name":
                company,

            "Current_Revenue_Bn":
                current_revenue / 1e9,

            "Previous_Revenue_Bn":
                previous_revenue_value / 1e9,

            "Revenue_YOY_%":
                revenue_yoy,

            "Current_Net_Income_Bn":
                current_net_income / 1e9,

            "Previous_Net_Income_Bn":
                previous_net_income_value / 1e9,

            "Net_Income_YOY_%":
                net_income_yoy,

            "Current_Margin_%":
                current_margin,

            "Previous_Margin_%":
                previous_margin,

            "Margin_Change_Points":
                margin_change,

            "Current_Period_Start":
                current_start,

            "Current_Period_End":
                current_end,

            "Previous_Period_Start":
                previous_revenue[
                    "start"
                ],

            "Previous_Period_End":
                previous_revenue[
                    "end"
                ]
        })


        print(
            f"✓ {company}"
        )


    except Exception as e:

        print(
            f"✗ {company}: {e}"
        )


# ============================================================
# CREATE DATAFRAME
# ============================================================

yoy_data = pd.DataFrame(
    yoy_results
)


# ============================================================
# ROUND NUMBERS
# ============================================================

if not yoy_data.empty:

    numeric_columns = [
        "Current_Revenue_Bn",
        "Previous_Revenue_Bn",
        "Revenue_YOY_%",
        "Current_Net_Income_Bn",
        "Previous_Net_Income_Bn",
        "Net_Income_YOY_%",
        "Current_Margin_%",
        "Previous_Margin_%",
        "Margin_Change_Points"
    ]

    yoy_data[
        numeric_columns
    ] = (
        yoy_data[
            numeric_columns
        ].round(2)
    )


# ============================================================
# DISPLAY
# ============================================================

yoy_data
```

```python
import pandas as pd
import requests

# ============================================================
# SEC SETTINGS
# ============================================================

headers = {
    "User-Agent": "Your Full Name your_email@example.com"
}


# ============================================================
# GET SEC COMPANY FACTS
# ============================================================

def get_company_facts(cik):

    url = (
        f"https://data.sec.gov/api/xbrl/companyfacts/"
        f"CIK{str(cik).zfill(10)}.json"
    )

    response = requests.get(
        url,
        headers=headers
    )

    response.raise_for_status()

    return response.json()


# ============================================================
# GET XBRL DATAFRAME
# ============================================================

def get_xbrl_df(facts, tag):

    try:

        units = facts["facts"]["us-gaap"][tag]["units"]

        if "USD" not in units:
            return None

        df = pd.DataFrame(
            units["USD"]
        )

        if df.empty:
            return None

        # Convert dates
        for col in ["start", "end", "filed"]:

            if col in df.columns:

                df[col] = pd.to_datetime(
                    df[col],
                    errors="coerce"
                )

        # Convert value
        df["val"] = pd.to_numeric(
            df["val"],
            errors="coerce"
        )

        # Calculate duration
        if (
            "start" in df.columns
            and
            "end" in df.columns
        ):

            df["duration_days"] = (
                df["end"] -
                df["start"]
            ).dt.days

        return df

    except Exception:

        return None


# ============================================================
# FIND REVENUE DATA
# ============================================================

def get_revenue_df(facts):

    revenue_tags = [
        "RevenueFromContractWithCustomerExcludingAssessedTax",
        "Revenues"
    ]

    for tag in revenue_tags:

        df = get_xbrl_df(
            facts,
            tag
        )

        if df is not None and not df.empty:

            df["Revenue_Tag"] = tag

            return df

    return None


# ============================================================
# FIND NET INCOME DATA
# ============================================================

def get_net_income_df(facts):

    net_income_tags = [
        "ProfitLoss",
        "NetIncomeLoss"
    ]

    for tag in net_income_tags:

        df = get_xbrl_df(
            facts,
            tag
        )

        if df is not None and not df.empty:

            df["Net_Income_Tag"] = tag

            return df

    return None


# ============================================================
# CLEAN QUARTERLY DATA
# ============================================================

def clean_quarterly(df):

    if df is None or df.empty:

        return pd.DataFrame()

    df = df.copy()

    # Only quarterly filings
    if "form" in df.columns:

        df = df[
            df["form"]
            .astype(str)
            .str.upper()
            .eq("10-Q")
        ]

    # Valid dates
    df = df[
        df["start"].notna()
        &
        df["end"].notna()
    ]

    # Quarterly duration
    if "duration_days" in df.columns:

        df = df[
            df["duration_days"].between(
                80,
                100
            )
        ]

    # Valid values
    df = df[
        df["val"].notna()
    ]

    return df.reset_index(
        drop=True
    )


# ============================================================
# FIND PRIOR-YEAR SAME QUARTER
#
# IMPORTANT:
#
# Current quarter is FIXED from automated_data.
#
# We are NOT looking for the latest SEC quarter.
#
# We match the prior year using actual dates.
#
# This is especially important for NVIDIA because its SEC
# comparative data can carry different FY labels.
# ============================================================

def find_prior_year_quarter(
    df,
    current_start,
    current_end
):

    df = clean_quarterly(df)

    if df.empty:

        return None

    current_start = pd.Timestamp(
        current_start
    )

    current_end = pd.Timestamp(
        current_end
    )

    current_duration = (
        current_end -
        current_start
    ).days

    # --------------------------------------------------------
    # Target dates approximately one year earlier
    # --------------------------------------------------------

    target_start = (
        current_start -
        pd.DateOffset(
            years=1
        )
    )

    target_end = (
        current_end -
        pd.DateOffset(
            years=1
        )
    )

    # --------------------------------------------------------
    # Calculate date distances
    # --------------------------------------------------------

    df["start_diff"] = (
        df["start"] -
        target_start
    ).abs().dt.days

    df["end_diff"] = (
        df["end"] -
        target_end
    ).abs().dt.days

    df["duration_diff"] = (
        df["duration_days"] -
        current_duration
    ).abs()

    df["date_diff"] = (
        df["start_diff"] +
        df["end_diff"]
    )

    # --------------------------------------------------------
    # Keep observations reasonably close to the expected
    # prior-year quarter.
    #
    # We deliberately DO NOT filter on fp or fy here.
    # --------------------------------------------------------

    candidates = df[
        (df["start_diff"] <= 60)
        &
        (df["end_diff"] <= 60)
        &
        (df["duration_diff"] <= 10)
    ].copy()

    if candidates.empty:

        return None

    # --------------------------------------------------------
    # Prefer:
    # 1. closest start date
    # 2. closest end date
    # 3. closest duration
    # 4. earliest filing
    # --------------------------------------------------------

    candidates = candidates.sort_values(
        [
            "date_diff",
            "duration_diff",
            "filed"
        ]
    )

    return candidates.iloc[0]


# ============================================================
# BUILD Q3 FY2025 vs Q3 FY2024 YOY ANALYSIS
# ============================================================

yoy_results = []


for _, row in automated_data.iterrows():

    company = row["name"]

    try:

        print(
            f"\nProcessing {company}..."
        )

        # ====================================================
        # CURRENT Q3 FY2025
        #
        # These values are taken directly from automated_data.
        # ====================================================

        current_start = pd.Timestamp(
            row["Period_Start"]
        )

        current_end = pd.Timestamp(
            row["Period_End"]
        )

        current_revenue = float(
            row["Revenue"]
        )

        current_net_income = float(
            row["Net_Income"]
        )


        # ====================================================
        # SEC FACTS
        # ====================================================

        cik = ciks[company]

        facts = get_company_facts(
            cik
        )


        # ====================================================
        # REVENUE
        # ====================================================

        revenue_df = get_revenue_df(
            facts
        )

        if revenue_df is None:

            print(
                f"✗ {company}: revenue unavailable"
            )

            continue


        # ====================================================
        # NET INCOME
        # ====================================================

        net_income_df = get_net_income_df(
            facts
        )

        if net_income_df is None:

            print(
                f"✗ {company}: net income unavailable"
            )

            continue


        # ====================================================
        # PRIOR-YEAR REVENUE
        # ====================================================

        prior_revenue_row = find_prior_year_quarter(
            revenue_df,
            current_start,
            current_end
        )

        if prior_revenue_row is None:

            print(
                f"✗ {company}: "
                f"Q3 FY2024 revenue not found"
            )

            continue


        # ====================================================
        # PRIOR-YEAR NET INCOME
        # ====================================================

        prior_net_income_row = find_prior_year_quarter(
            net_income_df,
            current_start,
            current_end
        )

        if prior_net_income_row is None:

            print(
                f"✗ {company}: "
                f"Q3 FY2024 net income not found"
            )

            continue


        # ====================================================
        # PRIOR-YEAR VALUES
        # ====================================================

        prior_revenue = float(
            prior_revenue_row["val"]
        )

        prior_net_income = float(
            prior_net_income_row["val"]
        )


        # ====================================================
        # MARGINS
        # ====================================================

        current_margin = (
            current_net_income /
            current_revenue
        ) * 100

        prior_margin = (
            prior_net_income /
            prior_revenue
        ) * 100


        # ====================================================
        # YOY GROWTH
        # ====================================================

        revenue_yoy = (
            (
                current_revenue /
                prior_revenue
            ) - 1
        ) * 100

        net_income_yoy = (
            (
                current_net_income /
                prior_net_income
            ) - 1
        ) * 100

        margin_change = (
            current_margin -
            prior_margin
        )


        # ====================================================
        # SAVE RESULTS
        # ====================================================

        yoy_results.append({

            "name":
                company,

            "Current_Revenue_Bn":
                current_revenue / 1e9,

            "Prior_Year_Revenue_Bn":
                prior_revenue / 1e9,

            "Revenue_YOY_%":
                revenue_yoy,

            "Current_Net_Income_Bn":
                current_net_income / 1e9,

            "Prior_Year_Net_Income_Bn":
                prior_net_income / 1e9,

            "Net_Income_YOY_%":
                net_income_yoy,

            "Current_Margin_%":
                current_margin,

            "Prior_Year_Margin_%":
                prior_margin,

            "Margin_Change_Points":
                margin_change,

            "Current_Period_Start":
                current_start,

            "Current_Period_End":
                current_end,

            "Prior_Year_Period_Start":
                prior_revenue_row["start"],

            "Prior_Year_Period_End":
                prior_revenue_row["end"]
        })


        # ====================================================
        # SHOW EXACT NVIDIA-STYLE PERIOD SELECTION
        # ====================================================

        print(
            f"✓ {company}"
        )

        print(
            f"   Current: "
            f"{current_start.date()} → "
            f"{current_end.date()} | "
            f"${current_revenue/1e9:.3f}B"
        )

        print(
            f"   Prior:   "
            f"{prior_revenue_row['start'].date()} → "
            f"{prior_revenue_row['end'].date()} | "
            f"${prior_revenue/1e9:.3f}B"
        )


    except Exception as e:

        print(
            f"✗ {company}: {e}"
        )


# ============================================================
# FINAL DATAFRAME
# ============================================================

yoy_data = pd.DataFrame(
    yoy_results
)


# ============================================================
# ROUND NUMBERS
# ============================================================

if not yoy_data.empty:

    numeric_columns = [

        "Current_Revenue_Bn",

        "Prior_Year_Revenue_Bn",

        "Revenue_YOY_%",

        "Current_Net_Income_Bn",

        "Prior_Year_Net_Income_Bn",

        "Net_Income_YOY_%",

        "Current_Margin_%",

        "Prior_Year_Margin_%",

        "Margin_Change_Points"
    ]

    yoy_data[numeric_columns] = (
        yoy_data[numeric_columns]
        .round(2)
    )


# ============================================================
# DISPLAY
# ============================================================

display(
    yoy_data
)
```

```python
import pandas as pd
import requests


# ============================================================
# SEC USER AGENT
# ============================================================

headers = {
    "User-Agent": "YOUR NAME YOUR_EMAIL@example.com"
}


# ============================================================
# GET SEC COMPANY FACTS
# ============================================================

def get_company_facts(cik):

    url = (
        f"https://data.sec.gov/api/xbrl/companyfacts/"
        f"CIK{str(cik).zfill(10)}.json"
    )

    response = requests.get(
        url,
        headers=headers
    )

    response.raise_for_status()

    return response.json()


# ============================================================
# CONVERT XBRL TAG INTO DATAFRAME
# ============================================================

def get_xbrl_df(facts, tag):

    try:

        tag_data = facts["facts"]["us-gaap"][tag]

        units = tag_data["units"]

        # We only need USD monetary values
        if "USD" not in units:
            return None

        df = pd.DataFrame(
            units["USD"]
        )

        if df.empty:
            return None

        # ----------------------------------------------------
        # Dates
        # ----------------------------------------------------

        for col in ["start", "end", "filed"]:

            if col in df.columns:

                df[col] = pd.to_datetime(
                    df[col],
                    errors="coerce"
                )

        # ----------------------------------------------------
        # Value
        # ----------------------------------------------------

        df["val"] = pd.to_numeric(
            df["val"],
            errors="coerce"
        )

        # ----------------------------------------------------
        # Duration
        # ----------------------------------------------------

        if (
            "start" in df.columns
            and
            "end" in df.columns
        ):

            df["duration_days"] = (
                df["end"]
                - df["start"]
            ).dt.days

        return df

    except Exception:

        return None


# ============================================================
# CLEAN QUARTERLY DATA
# ============================================================

def clean_quarterly(df):

    if df is None or df.empty:

        return pd.DataFrame()

    df = df.copy()

    # --------------------------------------------------------
    # Only 10-Q
    # --------------------------------------------------------

    if "form" in df.columns:

        df = df[
            df["form"]
            .astype(str)
            .str.upper()
            .eq("10-Q")
        ]

    # --------------------------------------------------------
    # Valid dates
    # --------------------------------------------------------

    df = df[
        df["start"].notna()
        &
        df["end"].notna()
    ]

    # --------------------------------------------------------
    # Quarterly duration
    # --------------------------------------------------------

    if "duration_days" in df.columns:

        df = df[
            df["duration_days"].between(
                80,
                100
            )
        ]

    # --------------------------------------------------------
    # Valid values
    # --------------------------------------------------------

    df = df[
        df["val"].notna()
    ]

    return df.reset_index(
        drop=True
    )


# ============================================================
# FIND REVENUE TAG THAT CONTAINS THE CURRENT QUARTER
#
# This is the important fix for NVIDIA.
#
# We do NOT assume that every company uses the same tag.
# ============================================================

def get_revenue_df(
    facts,
    current_start=None,
    current_end=None
):

    revenue_tags = [

        "RevenueFromContractWithCustomerExcludingAssessedTax",

        "Revenues",

        "SalesRevenueNet",

        "SalesRevenueGoodsNet",

        "SalesRevenueServicesNet"
    ]

    candidates = []

    for tag in revenue_tags:

        df = get_xbrl_df(
            facts,
            tag
        )

        if df is None or df.empty:

            continue

        df = clean_quarterly(
            df
        )

        if df.empty:

            continue

        df["Revenue_Tag"] = tag

        # ----------------------------------------------------
        # If current period is supplied, check whether this
        # tag actually contains that current quarter.
        # ----------------------------------------------------

        if (
            current_start is not None
            and
            current_end is not None
        ):

            current_start = pd.Timestamp(
                current_start
            )

            current_end = pd.Timestamp(
                current_end
            )

            start_diff = (
                df["start"]
                - current_start
            ).abs().dt.days

            end_diff = (
                df["end"]
                - current_end
            ).abs().dt.days

            matching_rows = df[
                (start_diff <= 3)
                &
                (end_diff <= 3)
            ]

            if not matching_rows.empty:

                # This tag actually contains the
                # current quarter.
                return df

        else:

            candidates.append(
                df
            )

    # --------------------------------------------------------
    # Fallback
    # --------------------------------------------------------

    if candidates:

        return candidates[0]

    return None


# ============================================================
# NET INCOME
# ============================================================

def get_net_income_df(facts):

    net_income_tags = [

        "ProfitLoss",

        "NetIncomeLoss",

        "NetIncomeLossAvailableToCommonStockholdersBasic"
    ]

    for tag in net_income_tags:

        df = get_xbrl_df(
            facts,
            tag
        )

        if df is None or df.empty:

            continue

        df = clean_quarterly(
            df
        )

        if not df.empty:

            df["Net_Income_Tag"] = tag

            return df

    return None


# ============================================================
# FIND SAME QUARTER PRIOR YEAR
#
# We compare PERIOD DATES, not FY / FP.
#
# Example NVIDIA:
#
# Current:
# 2024-07-29 → 2024-10-27
#
# Prior:
# 2023-07-31 → 2023-10-29
#
# ============================================================

def find_prior_year_quarter(
    df,
    current_start,
    current_end
):

    df = clean_quarterly(
        df
    )

    if df.empty:

        return None

    current_start = pd.Timestamp(
        current_start
    )

    current_end = pd.Timestamp(
        current_end
    )

    current_duration = (
        current_end -
        current_start
    ).days

    # --------------------------------------------------------
    # Expected prior-year dates
    # --------------------------------------------------------

    target_start = (
        current_start -
        pd.DateOffset(
            years=1
        )
    )

    target_end = (
        current_end -
        pd.DateOffset(
            years=1
        )
    )

    # --------------------------------------------------------
    # Date differences
    # --------------------------------------------------------

    df["start_diff"] = (
        df["start"]
        - target_start
    ).abs().dt.days

    df["end_diff"] = (
        df["end"]
        - target_end
    ).abs().dt.days

    df["duration_diff"] = (
        df["duration_days"]
        - current_duration
    ).abs()

    # --------------------------------------------------------
    # Find candidates
    # --------------------------------------------------------

    candidates = df[
        (df["start_diff"] <= 60)
        &
        (df["end_diff"] <= 60)
        &
        (df["duration_diff"] <= 10)
    ].copy()

    if candidates.empty:

        return None

    # --------------------------------------------------------
    # Rank by closeness of actual dates
    # --------------------------------------------------------

    candidates["date_diff"] = (
        candidates["start_diff"]
        +
        candidates["end_diff"]
    )

    candidates = candidates.sort_values(
        [
            "date_diff",
            "duration_diff",
            "filed"
        ]
    )

    # --------------------------------------------------------
    # Important:
    #
    # If the same period appears in multiple SEC filings,
    # select the earliest filing.
    # --------------------------------------------------------

    best_date_diff = (
        candidates.iloc[0]["date_diff"]
    )

    best_duration_diff = (
        candidates.iloc[0]["duration_diff"]
    )

    candidates = candidates[
        (candidates["date_diff"] == best_date_diff)
        &
        (
            candidates["duration_diff"]
            == best_duration_diff
        )
    ].sort_values(
        "filed"
    )

    return candidates.iloc[0]


# ============================================================
# MAIN YOY ANALYSIS
# ============================================================

yoy_results = []


for _, row in automated_data.iterrows():

    company = row["name"]

    try:

        print(
            f"\nProcessing {company}...."
        )

        # ====================================================
        # CURRENT QUARTER
        #
        # Comes directly from automated_data.
        # ====================================================

        current_start = pd.Timestamp(
            row["Period_Start"]
        )

        current_end = pd.Timestamp(
            row["Period_End"]
        )

        current_revenue = float(
            row["Revenue"]
        )

        current_net_income = float(
            row["Net_Income"]
        )


        # ====================================================
        # GET SEC FACTS
        # ====================================================

        cik = ciks[company]

        facts = get_company_facts(
            cik
        )


        # ====================================================
        # REVENUE
        #
        # We pass current dates so the function can determine
        # which revenue tag actually contains this quarter.
        # ====================================================

        revenue_df = get_revenue_df(
            facts,
            current_start,
            current_end
        )

        if revenue_df is None:

            print(
                f"✗ {company}: revenue data not found"
            )

            continue


        # ====================================================
        # NET INCOME
        # ====================================================

        net_income_df = get_net_income_df(
            facts
        )

        if net_income_df is None:

            print(
                f"✗ {company}: net income data not found"
            )

            continue


        # ====================================================
        # PRIOR-YEAR REVENUE
        # ====================================================

        prior_revenue_row = (
            find_prior_year_quarter(
                revenue_df,
                current_start,
                current_end
            )
        )

        if prior_revenue_row is None:

            print(
                f"✗ {company}: "
                f"prior-year revenue not found"
            )

            continue


        # ====================================================
        # PRIOR-YEAR NET INCOME
        # ====================================================

        prior_net_income_row = (
            find_prior_year_quarter(
                net_income_df,
                current_start,
                current_end
            )
        )

        if prior_net_income_row is None:

            print(
                f"✗ {company}: "
                f"prior-year net income not found"
            )

            continue


        # ====================================================
        # PRIOR-YEAR VALUES
        # ====================================================

        prior_revenue = float(
            prior_revenue_row["val"]
        )

        prior_net_income = float(
            prior_net_income_row["val"]
        )


        # ====================================================
        # MARGINS
        # ====================================================

        current_margin = (
            current_net_income
            /
            current_revenue
        ) * 100

        prior_margin = (
            prior_net_income
            /
            prior_revenue
        ) * 100


        # ====================================================
        # YOY GROWTH
        # ====================================================

        revenue_yoy = (
            (
                current_revenue
                /
                prior_revenue
            )
            - 1
        ) * 100

        net_income_yoy = (
            (
                current_net_income
                /
                prior_net_income
            )
            - 1
        ) * 100

        margin_change = (
            current_margin
            -
            prior_margin
        )


        # ====================================================
        # SAVE RESULT
        # ====================================================

        yoy_results.append({

            "name":
                company,

            "Current_Revenue_Bn":
                current_revenue / 1e9,

            "Prior_Year_Revenue_Bn":
                prior_revenue / 1e9,

            "Revenue_YOY_%":
                revenue_yoy,

            "Current_Net_Income_Bn":
                current_net_income / 1e9,

            "Prior_Year_Net_Income_Bn":
                prior_net_income / 1e9,

            "Net_Income_YOY_%":
                net_income_yoy,

            "Current_Margin_%":
                current_margin,

            "Prior_Year_Margin_%":
                prior_margin,

            "Margin_Change_Points":
                margin_change,

            "Current_Period_Start":
                current_start,

            "Current_Period_End":
                current_end,

            "Prior_Year_Period_Start":
                prior_revenue_row["start"],

            "Prior_Year_Period_End":
                prior_revenue_row["end"],

            "Revenue_Tag":
                prior_revenue_row.get(
                    "Revenue_Tag",
                    ""
                )
        })


        # ====================================================
        # VERIFICATION OUTPUT
        # ====================================================

        print(
            f"✓ {company}"
        )

        print(
            f"   Current: "
            f"{current_start.date()} → "
            f"{current_end.date()} | "
            f"${current_revenue / 1e9:.3f}B"
        )

        print(
            f"   Prior:   "
            f"{prior_revenue_row['start'].date()} → "
            f"{prior_revenue_row['end'].date()} | "
            f"${prior_revenue / 1e9:.3f}B"
        )

        print(
            f"   Revenue tag: "
            f"{revenue_df['Revenue_Tag'].iloc[0]}"
        )


    except Exception as e:

        print(
            f"✗ {company}: {e}"
        )


# ============================================================
# FINAL YOY DATAFRAME
# ============================================================

yoy_data = pd.DataFrame(
    yoy_results
)


# ============================================================
# ROUND NUMBERS
# ============================================================

if not yoy_data.empty:

    numeric_columns = [

        "Current_Revenue_Bn",

        "Prior_Year_Revenue_Bn",

        "Revenue_YOY_%",

        "Current_Net_Income_Bn",

        "Prior_Year_Net_Income_Bn",

        "Net_Income_YOY_%",

        "Current_Margin_%",

        "Prior_Year_Margin_%",

        "Margin_Change_Points"
    ]

    yoy_data[numeric_columns] = (
        yoy_data[numeric_columns]
        .round(2)
    )


# ============================================================
# DISPLAY FINAL RESULT
# ============================================================

display(
    yoy_data
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# STEP: YOY PERFORMANCE SCORE
# ============================================================

yoy_score = yoy_analysis.copy()


# ============================================================
# 1. RANK EACH YOY METRIC
# Higher = better
# ============================================================

yoy_score["Revenue_Growth_Rank"] = (
    yoy_score["Revenue_YOY_Growth_%"]
    .rank(
        ascending=False,
        method="min"
    )
)

yoy_score["Net_Income_Growth_Rank"] = (
    yoy_score["Net_Income_YOY_Growth_%"]
    .rank(
        ascending=False,
        method="min"
    )
)

yoy_score["Margin_Change_Rank"] = (
    yoy_score["Margin_Change_pp"]
    .rank(
        ascending=False,
        method="min"
    )
)


# ============================================================
# 2. CONVERT RANKS INTO SCORES
#
# Best company = 1
# Worst company = 0
# ============================================================

n = len(yoy_score)

if n > 1:

    yoy_score["Revenue_Growth_Score"] = (
        (n - yoy_score["Revenue_Growth_Rank"])
        / (n - 1)
    )

    yoy_score["Net_Income_Growth_Score"] = (
        (n - yoy_score["Net_Income_Growth_Rank"])
        / (n - 1)
    )

    yoy_score["Margin_Change_Score"] = (
        (n - yoy_score["Margin_Change_Rank"])
        / (n - 1)
    )

else:

    yoy_score["Revenue_Growth_Score"] = 1
    yoy_score["Net_Income_Growth_Score"] = 1
    yoy_score["Margin_Change_Score"] = 1


# ============================================================
# 3. EQUAL WEIGHTED YOY SCORE
# ============================================================

yoy_score["YOY_Performance_Score"] = (
      yoy_score["Revenue_Growth_Score"] * 0.333333
    + yoy_score["Net_Income_Growth_Score"] * 0.333333
    + yoy_score["Margin_Change_Score"] * 0.333334
)


# ============================================================
# 4. FINAL RANK
# ============================================================

yoy_score["YOY_Performance_Rank"] = (
    yoy_score["YOY_Performance_Score"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)


# ============================================================
# 5. SORT BY FINAL RANK
# ============================================================

yoy_score = yoy_score.sort_values(
    "YOY_Performance_Rank"
).reset_index(
    drop=True
)


# ============================================================
# 6. DISPLAY CLEAN RESULT
# ============================================================

yoy_score_display = yoy_score[
    [
        "name",
        "Revenue_YOY_Growth_%",
        "Net_Income_YOY_Growth_%",
        "Margin_Change_pp",
        "Revenue_Growth_Score",
        "Net_Income_Growth_Score",
        "Margin_Change_Score",
        "YOY_Performance_Score",
        "YOY_Performance_Rank"
    ]
].copy()


# ============================================================
# ROUND SCORES
# ============================================================

score_columns = [
    "Revenue_Growth_%",
    "Net_Income_Growth_%",
    "Margin_Change_pp",
    "Revenue_Growth_Score",
    "Net_Income_Growth_Score",
    "Margin_Change_Score",
    "YOY_Performance_Score"
]

for col in score_columns:

    if col in yoy_score_display.columns:

        yoy_score_display[col] = (
            yoy_score_display[col]
            .round(3)
        )


display(
    yoy_score_display
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# YOY QUALITY / MOMENTUM SCORE
# ============================================================

quality_score = yoy_score.copy()


# ============================================================
# 1. NORMALIZE THE THREE YOY METRICS
#
# Rank-based normalization:
# Best = 1
# Worst = 0
# ============================================================

n = len(quality_score)

quality_score["Revenue_Growth_Score"] = (
    quality_score["Revenue_YOY_Growth_%"]
    .rank(
        ascending=False,
        method="min"
    )
)

quality_score["Net_Income_Growth_Score"] = (
    quality_score["Net_Income_YOY_Growth_%"]
    .rank(
        ascending=False,
        method="min"
    )
)

quality_score["Margin_Change_Score"] = (
    quality_score["Margin_Change_pp"]
    .rank(
        ascending=False,
        method="min"
    )
)


# ============================================================
# CONVERT RANKS TO 0-1 SCORES
# ============================================================

if n > 1:

    quality_score["Revenue_Score"] = (
        (n - quality_score["Revenue_Growth_Score"])
        / (n - 1)
    )

    quality_score["Net_Income_Score"] = (
        (n - quality_score["Net_Income_Growth_Score"])
        / (n - 1)
    )

    quality_score["Margin_Score"] = (
        (n - quality_score["Margin_Change_Score"])
        / (n - 1)
    )

else:

    quality_score["Revenue_Score"] = 1
    quality_score["Net_Income_Score"] = 1
    quality_score["Margin_Score"] = 1


# ============================================================
# 2. WEIGHTED YOY QUALITY SCORE
#
# Revenue       = 30%
# Net Income    = 40%
# Margin        = 30%
# ============================================================

quality_score["YOY_Quality_Score"] = (
      quality_score["Revenue_Score"] * 0.30
    + quality_score["Net_Income_Score"] * 0.40
    + quality_score["Margin_Score"] * 0.30
)


# ============================================================
# 3. FINAL RANK
# ============================================================

quality_score["YOY_Quality_Rank"] = (
    quality_score["YOY_Quality_Score"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)


# ============================================================
# 4. SORT
# ============================================================

quality_score = quality_score.sort_values(
    "YOY_Quality_Rank"
).reset_index(
    drop=True
)


# ============================================================
# 5. CLEAN DISPLAY TABLE
# ============================================================

quality_display = quality_score[
    [
        "name",
        "Revenue_YOY_Growth_%",
        "Net_Income_YOY_Growth_%",
        "Margin_Change_pp",
        "Revenue_Score",
        "Net_Income_Score",
        "Margin_Score",
        "YOY_Quality_Score",
        "YOY_Quality_Rank"
    ]
].copy()


# ============================================================
# ROUND VALUES
# ============================================================

numeric_columns = [
    "Revenue_YOY_Growth_%",
    "Net_Income_YOY_Growth_%",
    "Margin_Change_pp",
    "Revenue_Score",
    "Net_Income_Score",
    "Margin_Score",
    "YOY_Quality_Score"
]

quality_display[numeric_columns] = (
    quality_display[numeric_columns]
    .round(3)
)


# ============================================================
# DISPLAY
# ============================================================

display(quality_display)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Q3 EARNINGS TRAJECTORY ANALYSIS
# ============================================================

trajectory = dashboard[
    [
        "name",
        "Revenue_YOY_Growth_%",
        "Net_Income_YOY_Growth_%",
        "Margin_Change_pp"
    ]
].copy()


# ============================================================
# 1. CREATE STANDARDIZED TRAJECTORY SCORES
# ============================================================

# Revenue growth
trajectory["Revenue_Growth_Score"] = (
    trajectory["Revenue_YOY_Growth_%"]
    .rank(
        pct=True
    )
)

# Net income growth
trajectory["Profit_Growth_Score"] = (
    trajectory["Net_Income_YOY_Growth_%"]
    .rank(
        pct=True
    )
)

# Margin change
trajectory["Margin_Improvement_Score"] = (
    trajectory["Margin_Change_pp"]
    .rank(
        pct=True
    )
)


# ============================================================
# 2. CALCULATE EARNINGS TRAJECTORY SCORE
# ============================================================

trajectory["Earnings_Trajectory_Score"] = (
      trajectory["Revenue_Growth_Score"] * 0.30
    + trajectory["Profit_Growth_Score"] * 0.40
    + trajectory["Margin_Improvement_Score"] * 0.30
)


# ============================================================
# 3. RANK
# ============================================================

trajectory["Trajectory_Rank"] = (
    trajectory["Earnings_Trajectory_Score"]
    .rank(
        ascending=False,
        method="min"
    )
    .astype(int)
)


# ============================================================
# 4. CLASSIFY TRAJECTORY
# ============================================================

def classify_trajectory(row):

    revenue = row["Revenue_YOY_Growth_%"]
    profit = row["Net_Income_YOY_Growth_%"]
    margin = row["Margin_Change_pp"]

    if (
        revenue > 0
        and profit > 0
        and margin > 0
    ):
        return "Accelerating"

    elif (
        revenue > 0
        and profit > 0
        and margin <= 0
    ):
        return "Profitable Growth"

    elif (
        profit > revenue
        and profit > 0
    ):
        return "Margin-Led / Mixed"

    else:
        return "Deteriorating"


trajectory["Trajectory"] = (
    trajectory.apply(
        classify_trajectory,
        axis=1
    )
)


# ============================================================
# 5. SORT
# ============================================================

trajectory = trajectory.sort_values(
    "Trajectory_Rank"
).reset_index(
    drop=True
)


# ============================================================
# 6. DISPLAY TABLE
# ============================================================

trajectory_display = trajectory[
    [
        "name",
        "Revenue_YOY_Growth_%",
        "Net_Income_YOY_Growth_%",
        "Margin_Change_pp",
        "Earnings_Trajectory_Score",
        "Trajectory_Rank",
        "Trajectory"
    ]
].copy()


trajectory_display[
    [
        "Revenue_YOY_Growth_%",
        "Net_Income_YOY_Growth_%",
        "Margin_Change_pp",
        "Earnings_Trajectory_Score"
    ]
] = trajectory_display[
    [
        "Revenue_YOY_Growth_%",
        "Net_Income_YOY_Growth_%",
        "Margin_Change_pp",
        "Earnings_Trajectory_Score"
    ]
].round(3)


print(
    "Q3 EARNINGS TRAJECTORY RANKING"
)

display(
    trajectory_display
)


# ============================================================
# 7. TRAJECTORY CHART
# ============================================================

plt.figure(
    figsize=(14, 8)
)


bars = plt.barh(
    trajectory["name"],
    trajectory["Earnings_Trajectory_Score"]
)


# Add score + trajectory

for bar, (_, row) in zip(
    bars,
    trajectory.iterrows()
):

    label = (
        f"{row['Earnings_Trajectory_Score']:.3f}"
        f" | {row['Trajectory']}"
    )

    plt.text(
        row["Earnings_Trajectory_Score"] + 0.01,
        bar.get_y() + bar.get_height() / 2,
        label,
        va="center",
        fontsize=9
    )


plt.title(
    "Q3 Earnings Trajectory Ranking",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Earnings Trajectory Score"
)

plt.ylabel(
    "Company"
)

plt.xlim(
    0,
    1.05
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# FINAL RANKING ROBUSTNESS ANALYSIS
# ============================================================

# ------------------------------------------------------------
# 1. PREVIOUS 50/50 RANKING
# ------------------------------------------------------------

previous_ranking = pd.DataFrame({
    "name": [
        "ALPHABET INC.",
        "NVIDIA CORP",
        "MICROSOFT CORP",
        "WALMART INC.",
        "APPLE INC.",
        "META PLATFORMS, INC."
    ],
    "Previous_50_50_Rank": [
        1,
        2,
        3,
        4,
        5,
        6
    ]
})


# ------------------------------------------------------------
# 2. GET FINAL RANKING
# ------------------------------------------------------------

movement = final_ranking[
    [
        "name",
        "Final_Rank",
        "Final_Score"
    ]
].copy()


# ------------------------------------------------------------
# 3. MERGE OLD AND NEW RANKINGS
# ------------------------------------------------------------

movement = movement.merge(
    previous_ranking,
    on="name",
    how="left"
)


# ------------------------------------------------------------
# 4. CALCULATE RANK MOVEMENT
# ------------------------------------------------------------

movement["Rank_Change"] = (
    movement["Previous_50_50_Rank"]
    - movement["Final_Rank"]
)


# Positive = moved UP
# Negative = moved DOWN
# Zero = unchanged


# ------------------------------------------------------------
# 5. ADD MOVEMENT LABEL
# ------------------------------------------------------------

def movement_label(value):

    if value > 0:
        return "Moved Up"

    elif value < 0:
        return "Moved Down"

    else:
        return "Unchanged"


movement["Movement"] = (
    movement["Rank_Change"]
    .apply(movement_label)
)


# ------------------------------------------------------------
# 6. SORT BY FINAL RANK
# ------------------------------------------------------------

movement = movement.sort_values(
    "Final_Rank"
).reset_index(
    drop=True
)


# ------------------------------------------------------------
# 7. DISPLAY RESULT
# ------------------------------------------------------------

movement_display = movement[
    [
        "Final_Rank",
        "name",
        "Previous_50_50_Rank",
        "Final_Score",
        "Rank_Change",
        "Movement"
    ]
].copy()


movement_display.columns = [
    "Final Rank",
    "Company",
    "Previous 50/50 Rank",
    "Final Score",
    "Rank Change",
    "Movement"
]


print(
    "FINAL RANKING ROBUSTNESS ANALYSIS"
)

display(movement_display)


# ============================================================
# 8. CHECK WHETHER RANKINGS ARE IDENTICAL
# ============================================================

rankings_identical = (
    movement["Previous_50_50_Rank"]
    == movement["Final_Rank"]
).all()


print(
    "\nRanking Robustness:"
)

if rankings_identical:

    print(
        "✓ Final ranking is identical to the previous 50/50 ranking."
    )

    print(
        "✓ No company changed position."
    )

else:

    print(
        "⚠ Final ranking changed after adding Earnings Trajectory."
    )


# ============================================================
# 9. CREATE RANK MOVEMENT CHART
# ============================================================

plt.figure(
    figsize=(12, 7)
)


bars = plt.barh(
    movement["name"],
    movement["Rank_Change"]
)


# ------------------------------------------------------------
# ADD MOVEMENT VALUES
# ------------------------------------------------------------

for bar, value in zip(
    bars,
    movement["Rank_Change"]
):

    if value > 0:

        label = f"+{value}"

    else:

        label = str(value)


    if value >= 0:

        x_position = value + 0.05

    else:

        x_position = value - 0.15


    plt.text(
        x_position,
        bar.get_y() + bar.get_height() / 2,
        label,
        va="center",
        fontsize=10
    )


# ------------------------------------------------------------
# ZERO LINE
# ------------------------------------------------------------

plt.axvline(
    0,
    linewidth=1
)


plt.title(
    "Final Ranking Movement vs Previous 50/50 Model",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel(
    "Rank Change"
)

plt.ylabel(
    "Company"
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.tight_layout()

plt.show()
```
